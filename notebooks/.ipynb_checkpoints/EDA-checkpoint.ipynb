# Exploratory Data Analysis (EDA) of Akkadian Cuneiform Dataset
This notebook analyzes the `CuneiMLv1.2` dataset.

```python
import json
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
import numpy as np

DATA_PATH = "../data/prepared_datasets/train.jsonl"

print(f"Loading data from {DATA_PATH}...")
data = []
with open(DATA_PATH, "r", encoding="utf-8") as f:
    for line in f:
        data.append(json.loads(line))

print(f"Total lines loaded: {len(data)}")
```

```python
# 1. Analysis of Signs (Characters)
all_signs = []
sequence_lengths = []
lacuna_count = 0

for row in data:
    signs = row["signs"]
    all_signs.extend(signs)
    sequence_lengths.append(len(signs))

    # Check for missing signs (x, ..., etc)
    raw = row["raw"].lower()
    if "x" in raw or "..." in raw or "[" in raw or "]" in raw:
        lacuna_count += 1

sign_counts = Counter(all_signs)
unique_signs = len(sign_counts)

print(f"Total signs in dataset: {len(all_signs)}")
print(f"Unique signs: {unique_signs}")
print(
    f"Lines with damage/lacunas: {lacuna_count} ({(lacuna_count / len(data)) * 100:.1f}%)"
)
```

```python
# 2. Plot Top 20 Most Frequent Signs
top_20 = sign_counts.most_common(20)
labels = [x[0] for x in top_20]
values = [x[1] for x in top_20]

plt.figure(figsize=(12, 6))
plt.bar(labels, values, color="skyblue")
plt.title("Top 20 Most Frequent Akkadian Signs")
plt.xlabel("Sign Transliteration")
plt.ylabel("Frequency")
plt.xticks(rotation=45)
plt.show()
```

```python
# 3. Sequence Length Distribution
plt.figure(figsize=(10, 5))
plt.hist(sequence_lengths, bins=50, color="lightgreen", edgecolor="black")
plt.title("Distribution of Sequence Lengths (Signs per line)")
plt.xlabel("Number of Signs")
plt.ylabel("Number of Lines")
plt.xlim(0, 50)  # Limit to 50 for better readability as most are short
plt.show()

print(f"Average signs per line: {np.mean(sequence_lengths):.1f}")
print(f"Max signs in a line: {np.max(sequence_lengths)}")
```
